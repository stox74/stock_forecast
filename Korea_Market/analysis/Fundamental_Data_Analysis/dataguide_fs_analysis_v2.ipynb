## 7. 개별 종목 재무 시계열 Excel 저장

티커와 재무항목(§1의 항목명 그대로)을 넣으면 DB에서 추출해
`FS_ts_data` 폴더에 `티커_항목_날짜.xlsx` 로 저장합니다.

```python
# ==========================================================
# PART 7: 개별 종목 재무 시계열 Excel 저장
# ==========================================================
from datetime import datetime

EXPORT_DIR = Path(r"C:\Users\82108\OneDrive\INVESTMENT\한국주식\FS_ts_data")


def export_fs_ts(ticker, items, start=None, unit=UNIT, out_dir=EXPORT_DIR):
    """
    개별 종목의 재무 시계열을 DB에서 추출해 Excel 로 저장한다.

    Parameters
    ----------
    ticker : str
        예: 'A017960' (A 접두사 형식)
    items : str 또는 list
        재무항목명. 예: '매출액' 또는 ['매출액', '영업이익']
        사용 가능한 항목명은 §1 의 A.search_items(engine) 로 확인.
    start : str, optional
        시작일. 예: '2018-01-01'. None 이면 전체 기간.
    unit : float
        단위 변환 제수. 기본 UNIT(1e5, 천원 -> 억원).
    out_dir : Path
        저장 폴더. 없으면 생성.

    Returns
    -------
    Path 또는 None (데이터 없음)
        저장된 파일 경로. 파일명: {ticker}_{항목}_{YYYYMMDD}.xlsx
        같은 날 같은 이름이 이미 있으면 _2, _3 으로 번호를 올려 저장.
    """
    if isinstance(items, str):
        items = [items]

    ts = A.get_ts(engine, ticker, items, start=start, unit=unit)
    if ts is None or ts.empty:
        print(f"데이터 없음: {ticker} / {items} — 티커 형식(A접두사)과 항목명을 확인하세요")
        return None

    out = ts.copy()
    out.index = out.index.astype(str)
    out.index.name = "분기"

    out_dir = Path(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    label = "_".join(items) if len(items) <= 3 else f"{items[0]}외{len(items) - 1}건"
    today = datetime.now().strftime("%Y%m%d")
    path = out_dir / f"{ticker}_{label}_{today}.xlsx"
    n = 2
    while path.exists():
        path = out_dir / f"{ticker}_{label}_{today}_{n}.xlsx"
        n += 1

    unit_label = {1e5: "억원", 1e3: "백만원", 1: "천원"}.get(unit, f"천원/{unit:g}")
    meta = pd.DataFrame(
        {"항목": ["ticker", "재무항목", "기간", "단위", "추출일시", "출처", "attrs"],
         "값": [ticker,
               ", ".join(items),
               f"{out.index[0]} ~ {out.index[-1]} ({len(out)}개 분기)",
               unit_label,
               datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
               "investar.korea_fs_data_from_DG (DataGuide)",
               str(ts.attrs)]}
    )

    with pd.ExcelWriter(path, engine="openpyxl") as xw:
        out.to_excel(xw, sheet_name="시계열")
        meta.to_excel(xw, sheet_name="메타", index=False)

    print(f"저장 완료: {path}")
    print(f"  {len(out)}개 분기 x {len(items)}개 항목, 단위: {unit_label}")
    return path
```
